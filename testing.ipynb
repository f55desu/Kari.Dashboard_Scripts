```python
# --- фильтрация по нужным датам ---
dates = ['23.10.2025', '22.10.2025', '21.10.2025', '20.10.2025']
df_filtered = df[df['Дата'].isin(dates)]

# --- группировка и агрегация ---
df_result = (
    df_filtered
    .groupby('Дата', as_index=False)
    .agg({
        'КолвоЗаказанных_распред': 'sum',
        'СуммаЗаказов_распред': 'sum'
    })
    .rename(columns={
        'КолвоЗаказанных_распред': 'CountOrdered',
        'СуммаЗаказов_распред': 'SummaryOrdered'
    })
)

print(df_result)
```
